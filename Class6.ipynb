```python
# Text analyzer
sentence = input("Please enter a sentence: ")
print(f"Number of characters: {len(sentence)} \n{"="*30}")
print(f"Number of words: {len(sentence.split())}\n{"="*30}")
print(f"Sentence in uppercase: {sentence.upper()}\n{"="*30}")
print(f"Sentence in reverse: {sentence[::-1]}\n{"="*30}")
for i in sentence.split():
  if(i.lower() == "python"):
    print(f"Does it contain 'python'? yes")
    break
  else:
    print(f"Does it contain 'python'? no")

```
