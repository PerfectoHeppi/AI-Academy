# 🧭 Жишээ даалгавар (бүрэн бөглөсөн загвар)

Доорх жишээ нь даалгаврын бүтэц хэрхэн ажилладгийг харуулна. Үүнийг **гүйцээх шаардлагагүй** — зөвхөн уншиж, ажиллуулж үзнэ үү.

### 🎯 Зорилго
`ages` толиос **18-аас дээш** насны хүмүүсийн нэрийг хэвлэнэ.

### 📥 Өгөгдөл
```python
ages = {"Bat": 17, "Sara": 21, "Max": 19}
```

### 🛠 Хийх алхмууд
1. Толийг `.items()`-ээр давтана.
2. Нас нь 18-аас их эсэхийг `if`-ээр шалгана.
3. Нөхцөл биелвэл нэрийг хэвлэнэ.

### ✅ Хүлээгдэх гаралт
```
Sara (21)
Max (19)
```

````python
# Жишээ — бүрэн бөглөсөн шийдэл
ages = {"Bat": 17, "Sara": 21, "Max": 19}

# 1) толийг .items()-ээр давтах
for name, age in ages.items():
    # 2) нөхцөл шалгах
    if age >= 18:
        # 3) хэвлэх
        print(name, "(" + str(age) + ")")
````

## 🟢 Даалгавар 1 — Сайн суралцаж буй суралцагчид

**Түвшин:** Хялбар  ·  **Ашиглах ойлголт:** `.items()` `if` `in`

### 🎯 Зорилго
`grades` толиос **A эсвэл B** дүнтэй суралцагчдын нэр, дүнг хэвлэнэ.

### 📥 Өгөгдөл
```python
grades = {"Alex": "A", "Sara": "B", "Max": "C", "Eva": "F", "Bat": "A"}
```

### 🛠 Хийх алхмууд
1. `grades` толийг `for name, grade in grades.items():` давталтаар гүйлгэнэ.
2. Дүн нь `["A", "B"]` жагсаалтад байгаа эсэхийг `if` -ээр шалгана.
3. Нөхцөл биелвэл `print(name, ":", grade)` гэж хэвлэнэ.

### ✅ Хүлээгдэх гаралт
```
Alex : A
Sara : B
Bat : A
```

> 💡 **Санамж:** `if grade in ["A", "B"]:` гэж бичвэл хоёр нөхцөлийг нэг мөрөөр шалгана.

````python
grades = {"Alex": "A", "Sara": "B", "Max": "C", "Eva": "F", "Bat": "A"}
for name , grade in grades.items():
    if (grade == "A" or grade == "B"):
        print(name, "(" + str(grade) + ")")

# 1) grades толийг .items()-ээр давтах

# 2) дүн нь "A" эсвэл "B" эсэхийг шалгах

# 3) тохирвол нэр, дүнг хэвлэх
````

## 🟢 Даалгавар 2 — Үсгийн давтамж

**Түвшин:** Хялбар  ·  **Ашиглах ойлголт:** `for` `if / else` `counts[ch] += 1`

### 🎯 Зорилго
Өгөгдсөн үгэн дэх **үсэг бүр хэдэн удаа** орсныг тоолж толь үүсгэнэ.

### 📥 Өгөгдөл
```python
word = "programming"
counts = {}
```

### 🛠 Хийх алхмууд
1. `for ch in word:` давталтаар үсэг бүрийг гүйлгэнэ.
2. Тухайн үсэг `counts` толинд **аль хэдийн байгаа** эсэхийг `if ch in counts:` -ээр шалгана.
3. Байвал `counts[ch] += 1`, эс байвал `counts[ch] = 1` гэж ононо.
4. Давталт дуусахад `counts` толийг хэвлэнэ.

### ✅ Хүлээгдэх гаралт
```
{'p': 1, 'r': 2, 'o': 1, 'g': 2, 'a': 1, 'm': 2, 'i': 1, 'n': 1}
```

> 💡 **Санамж:** Эхний удаа тааралдсан үсгийн утга байхгүй тул шууд `+= 1` хийвэл `KeyError` гарна.

````python
word = "programming"
counts = {}

# 1) word-ийн үсэг бүрийг давтах
for x in word:
    if x in counts:
        counts[x] += 1
    else:
        counts[x] = 1
# 2) үсэг counts дотор байгаа эсэхийг шалгах

# 3) байвал 1 нэмэх, эс байвал 1 гэж оноох

print(counts)
````

## 🟢 Даалгавар 3 — Үгийн давтамж

**Түвшин:** Хялбар  ·  **Ашиглах ойлголт:** `.lower()` `.split()` `for` `if / else`

### 🎯 Зорилго
Өгүүлбэр дэх **үг бүр хэдэн удаа** давтагдсаныг тоолж толь үүсгэнэ.

### 📥 Өгөгдөл
```python
text = "the cat and the dog and the bird"
word_count = {}
```

### 🛠 Хийх алхмууд
1. `text.lower()` -ээр бүх үсгийг жижиг болгоно.
2. `.split()` -ээр өгүүлбэрийг үг болгон салгана.
3. 2-р даалгаврын нэгэн адил үг бүрийн давтамжийг тоолно.
4. `word_count` толийг хэвлэнэ.

### ✅ Хүлээгдэх гаралт
```
{'the': 3, 'cat': 1, 'and': 2, 'dog': 1, 'bird': 1}
```

> 💡 **Санамж:** `for w in text.lower().split():` гэж хоёр үйлдлийг нэг мөрөнд хийж болно.

````python
text = "the cat and the dog and the bird"
word_count = {}
for x in text.lower().split():
    if x in word_count: 
        word_count[x] += 1
    else:
        word_count[x] = 1


# 1) text-ийг жижиг үсэг болгож, үг болгон салгах

# 2) үг бүрийн давтамжийг тоолох

print(word_count)
````

## 🔵 Даалгавар 4 — Өгөгдлийн стандартчлал

**Түвшин:** Дунд  ·  **Ашиглах ойлголт:** `.items()` `.title()` `.upper()`

### 🎯 Зорилго
`students` толийн **түлхүүр**-ийг `.title()`, **утга**-г `.upper()` хэлбэрт шилжүүлсэн **шинэ толь** байгуулна.

### 📥 Өгөгдөл
```python
students = {"bat-erdene": "ulaanbaatar", "saraa": "darkhan", "temuulen": "erdenet"}
clean = {}
```

### 🛠 Хийх алхмууд
1. `students` толийг `.items()` -ээр давтана.
2. Нэр дээр `.title()` хэрэглэж шинэ түлхүүр бэлдэнэ.
3. Хот дээр `.upper()` хэрэглэж шинэ утга бэлдэнэ.
4. `clean[шинэ_түлхүүр] = шинэ_утга` гэж оноон толийг цуглуулна.

### ✅ Хүлээгдэх гаралт
```
{'Bat-Erdene': 'ULAANBAATAR', 'Saraa': 'DARKHAN', 'Temuulen': 'ERDENET'}
```

> 💡 **Санамж:** Анхны `students` толийг өөрчлөхгүй — үр дүнг зөвхөн `clean` толинд хадгална.

````python
students = {"bat-erdene": "ulaanbaatar", "saraa": "darkhan", "temuulen": "erdenet"}
clean = {}
for x,y in students.items():
    clean[x.title()] = y.upper()
# 1) students толийг .items()-ээр давтах

# 2) нэрийг .title(), хотыг .upper() болгох

# 3) үр дүнг clean толинд нэмэх

print(clean)
````

## 🔵 Даалгавар 5 — Хамгийн урт түлхүүр

**Түвшин:** Дунд  ·  **Ашиглах ойлголт:** `for` `len()` `if`

### 🎯 Зорилго
`phones` толиос **хамгийн урт нэртэй** бүртгэлийг олж, нэр болон дугаарыг нь хэвлэнэ.

### 📥 Өгөгдөл
```python
phones = {"Bat": "99112233", "Enkhjargal": "88445566", "Sara": "95001122"}
longest = ""
```

### 🛠 Хийх алхмууд
1. `longest` хувьсагчийг хоосон мөрөөр эхлүүлнэ (аль хэдийн өгөгдсөн).
2. `for name in phones:` давталтаар түлхүүр бүрийг гүйлгэнэ.
3. `if len(name) > len(longest):` нөхцөлөөр урт эсэхийг шалгана.
4. Илүү урт бол `longest = name` гэж шинэчилнэ.
5. Давталт дуусахад `longest` болон `phones[longest]` -ийг хэвлэнэ.

### ✅ Хүлээгдэх гаралт
```
Хамгийн урт нэр: Enkhjargal -> 88445566
```

> 💡 **Санамж:** Толийг шууд `for name in phones:` гэж давтахад **түлхүүрүүд** гарч ирнэ.

````python
phones = {"Bat": "99112233", "Enkhjargal": "88445566", "Sara": "95001122"}
longest = ""
for name in phones:
    if len(name) > len(longest):
        longest = name
print(f"{longest} -> {phones[longest]}")

# 1) phones толийн түлхүүр бүрийг давтах

# 2) len() ашиглан longest-ээс урт эсэхийг шалгах

# 3) урт бол longest-ийг шинэчлэх

# 4) longest болон түүний дугаарыг хэвлэх
````

## 🔵 Даалгавар 6 — Утасны лавлах систем

**Түвшин:** Дунд  ·  **Ашиглах ойлголт:** `while True` `input()` `break` `in`

### 🎯 Зорилго
Хэрэглэгчээс нэр асууж, `contacts` толинд байвал дугаарыг, эс байвал `"Олдсонгүй"` гэж хэвлэнэ. `quit` гэж оруулах хүртэл давтана.

### 📥 Өгөгдөл
```python
contacts = {"Bat": "99112233", "Sara": "95001122", "Max": "88997766"}
```

### 🛠 Хийх алхмууд
1. `while True:` төгсгөлгүй давталт эхлүүлнэ (аль хэдийн бичигдсэн).
2. `input()` -ээр нэрийг авна (аль хэдийн бичигдсэн).
3. Хэрэв нэр нь `"quit"` бол `break` -ээр давталтаас гарна.
4. Хэрэв нэр нь `contacts` толинд байвал дугаарыг хэвлэнэ.
5. Байхгүй бол `"Олдсонгүй"` гэж хэвлэнэ.

### ✅ Хүлээгдэх гаралт
```
Нэр оруулна уу: Sara
Sara -> 95001122
Нэр оруулна уу: Doljin
Олдсонгүй
Нэр оруулна уу: quit
```

> 💡 **Санамж:** `break` -ийг эхэнд нь шалгахгүй бол давталт хэзээ ч зогсохгүй.

````python
contacts = {"Bat": "99112233", "Sara": "95001122", "Max": "88997766"}

while True:
    name = input("Нэр оруулна уу (гарахдаа quit): ")
    if(name == "quit"):
        break
    if name in contacts:
        print(f"{name} -> {contacts[name]}")
    else:
        print("Oldsongui")
    # 1) name == "quit" бол break

    # 2) name толинд байвал дугаарыг хэвлэх

    # 3) байхгүй бол "Олдсонгүй" гэж хэвлэх
````

## 🔴 Даалгавар 7 — Дундаж оноо ба үсгэн үнэлгээ

**Түвшин:** Ахисан  ·  **Ашиглах ойлголт:** `if / elif / else` `.items()` `len()`

### 🎯 Зорилго
Суралцагч бүрийн онооноос үсгэн үнэлгээ тодорхойлж, бүлгийн **дундаж оноог** тооцно.

### 📥 Өгөгдөл
```python
scores = {"Alex": 92, "Sara": 85, "Max": 71, "Eva": 58}
```

### 🛠 Хийх алхмууд
1. `total = 0` гэж нийлбэрийн хувьсагч бэлдэнэ.
2. `scores` толийг `.items()` -ээр давтана.
3. `if / elif / else` -ээр үнэлгээ тодорхойлно: 90+ = A, 80+ = B, 70+ = C, бусад = F.
4. Нэр, оноо, үнэлгээг хэвлээд `total` дээр оноог нэмнэ.
5. Давталт дуусахад `total / len(scores)` дунджийг хэвлэнэ.

### ✅ Хүлээгдэх гаралт
```
Alex : 92 -> A
Sara : 85 -> B
Max : 71 -> C
Eva : 58 -> F
Дундаж: 76.5
```

> 💡 **Санамж:** `elif` -ийг **буурах** дарааллаар (90 → 80 → 70) бичихгүй бол үнэлгээ буруу гарна.

````python
scores = {"Alex": 92, "Sara": 85, "Max": 71, "Eva": 58}
total = 0
for x in scores:
    total += scores[x]
    if(scores[x] >= 90):
        print(f"{x} : {scores[x]} -> A")
    elif(scores[x]>= 80):
        print(f"{x} : {scores[x]} -> B")
    elif(scores[x]>= 70):
        print(f"{x} : {scores[x]} -> C")   
    elif(scores[x]>= 60):
        print(f"{x} : {scores[x]} -> D")
    else:
        print(f"{x} : {scores[x]} -> F")
print(f"{total/len(scores)}")
# 1) scores толийг .items()-ээр давтах

# 2) if / elif / else-ээр үсгэн үнэлгээ тодорхойлох

# 3) нэр, оноо, үнэлгээг хэвлэх, total дээр оноог нэмэх

# 4) дундаж оноог хэвлэх
````

## 🔴 Даалгавар 8 — Урвуу толь байгуулах

**Түвшин:** Ахисан  ·  **Ашиглах ойлголт:** `.items()` `not in` `if`

### 🎯 Зорилго
`capitals` толийн **түлхүүр ба утгыг сольсон** шинэ толь байгуулна.

### 📥 Өгөгдөл
```python
capitals = {"Mongolia": "Ulaanbaatar", "Japan": "Tokyo", "Korea": "Seoul"}
reversed_dict = {}
```

### 🛠 Хийх алхмууд
1. `capitals` толийг `.items()` -ээр давтана.
2. Хот нь `reversed_dict` -д **байхгүй** эсэхийг `if city not in reversed_dict:` -ээр шалгана.
3. Байхгүй бол `reversed_dict[city] = country` гэж ононо.
4. `reversed_dict` толийг хэвлэнэ.

### ✅ Хүлээгдэх гаралт
```
{'Ulaanbaatar': 'Mongolia', 'Tokyo': 'Japan', 'Seoul': 'Korea'}
```

> 💡 **Санамж:** Утга давхардвал сүүлийнх нь өмнөхийг дарж бичдэг тул `not in` шалгалт хэрэгтэй.

````python
capitals = {"Mongolia": "Ulaanbaatar", "Japan": "Tokyo", "Korea": "Seoul"}
reversed_dict = {}
for x,y in capitals.items():
    if(x not in reversed_dict):
        reversed_dict[y] = x

# 1) capitals толийг .items()-ээр давтах

# 2) хот reversed_dict-д байхгүй эсэхийг шалгах

# 3) байхгүй бол хот -> улс гэж оноох

print(reversed_dict)
````

## 🔴 Даалгавар 9 — Тэмдэгт орлуулах шифр

**Түвшин:** Ахисан  ·  **Ашиглах ойлголт:** `for` `in` `secret += ch`

### 🎯 Зорилго
`cipher` толийг ашиглан өгүүлбэрийн үсэг бүрийг орлуулж **шинэ тэмдэгт мөр** үүсгэнэ.

### 📥 Өгөгдөл
```python
cipher = {"a": "@", "e": "3", "i": "1", "o": "0", "s": "5"}
message = "ai academy is nice"
secret = ""
```

### 🛠 Хийх алхмууд
1. `for ch in message:` давталтаар тэмдэгт бүрийг гүйлгэнэ.
2. Тухайн тэмдэгт `cipher` толинд байгаа эсэхийг `if ch in cipher:` -ээр шалгана.
3. Байвал `secret += cipher[ch]`, эс байвал `secret += ch` гэж нэмнэ.
4. `secret` мөрийг хэвлэнэ.

### ✅ Хүлээгдэх гаралт
```
@1 @c@d3my 15 n1c3
```

> 💡 **Санамж:** Зай (space) толинд байхгүй тул хэвээрээ үлдэж, үг тус тусдаа хэвээр харагдана.

````python
cipher = {"a": "@", "e": "3", "i": "1", "o": "0", "s": "5"}
message = "ai academy is nice"
secret = ""
for x in message:
    if x in cipher.keys():
        secret += cipher[x]
    else:
        secret += x 

# 1) message-ийн тэмдэгт бүрийг давтах

# 2) тэмдэгт cipher толинд байгаа эсэхийг шалгах

# 3) байвал орлуулсан, эс байвал анхны тэмдэгтийг secret-д нэмэх

print(secret)
````
