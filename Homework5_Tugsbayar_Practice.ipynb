````python
def add_up(numbers):
    total = 0
    for n in numbers:
        total = total + n
        print("Одоогийн нийлбэр:", total)
    print("Эцсийн дүн:", total)


add_up([10, 20, 30])
````

### B5 — Үнэлгээний үсэг

`print_grades(scores)` функц бичээрэй. Оноо бүрийг дараах хүснэгтээр ангилж хэвлэнэ.

| Оноо | Үсэг |
|------|------|
| 90+ | A |
| 80–89 | B |
| 70–79 | C |
| 60–69 | D |
| 60-аас доош | F |

```python
print_grades([95, 83, 71, 65, 40])
```

**Хүлээгдэх үр дүн:**
```
95 → A
83 → B
71 → C
65 → D
40 → F
```

> 💡 `if / elif / elif / elif / else` гинжийг ашиглана. Дараалал нь чухал — **их утгаас нь эхэл**.

````python
# Энд бичнэ үү
def print_grades(scores):
    for score in scores:
        if score >= 90:
            print(f"{score}: A")
        elif score >= 80:
            print(f"{score}: B")
        elif score >= 70:
            print(f"{score}: C")
        elif score >= 60:
            print(f"{score}: D")
        else:
            print(f"{score}: F")
print_grades([95, 83, 71, 65, 40])
````

### B7 — Хайлт (флаг хувьсагч)

`search_item(items, target)` функц бичээрэй. List дотор тухайн зүйл байгаа эсэхийг хэвлэнэ.

```python
search_item(["алим", "гадил", "жүрж"], "гадил")
search_item(["алим", "гадил", "жүрж"], "усан үзэм")
```

**Хүлээгдэх үр дүн:**
```
гадил → олдлоо ✅
усан үзэм → олдсонгүй ❌
```

> 💡 **Санаа:** `return` ашиглахгүй тул **флаг хувьсагч** хэрэглэнэ:
> 1. Loop-ийн өмнө `found = False`
> 2. Олдвол `found = True`
> 3. Loop дууссаны **дараа** `found`-ыг шалгаж хэвлэнэ

````python
# Энд бичнэ үү
def search_item(items, target):
    found = False   
    for item in items:
        if item == target:
            found = True
    if(found == True):
        print(f"{target}: found")
    else:
        print(f"{target}: not found")
search_item(["алим", "гадил", "жүрж"], "жүрж")
search_item(["алим", "гадил", "жүрж"], "усан үзэм")
````

### B8 — Статистик (хуримтлуулагч)

`print_stats(numbers)` функц бичээрэй. Дараах 4 зүйлийг хэвлэнэ: **тоо ширхэг, нийлбэр, дундаж, хамгийн их**.

```python
print_stats([10, 25, 8, 42, 15])
```

**Хүлээгдэх үр дүн:**
```
Тоо ширхэг: 5
Нийлбэр: 100
Дундаж: 20.0
Хамгийн их: 42
```

> 💡 A5 дасгалын **хуримтлуулагч** загварыг ашигла. Хамгийн ихийг олохдоо: эхний элементийг "одоогийн аварга" болгож аваад, түүнээс их тоо тааралдвал сольсоор явна.

````python
# Энд бичнэ үү
def print_stats(numbers):
    total = 0
    count = 0
    for n in numbers:
        total += n
        count += 1
    average = total / count if count > 0 else 0
    print(f"Count: {count}")
    print(f"Total: {total}")
    print(f"Average: {average}")
    print(f"Max: {max(numbers)}")
print_stats([10, 25, 8, 42, 15])
````

### B9 — Хоёр list зэрэг

`print_price_check(products, prices, limit)` функц бичээрэй. Бараа бүрийн үнийг хязгаартай харьцуулна.

```python
products = ["сүү", "талх", "мах", "давс"]
prices = [3500, 2000, 25000, 1500]
print_price_check(products, prices, 5000)
```

**Хүлээгдэх үр дүн:**
```
сүү - 3500₮ (хямд)
талх - 2000₮ (хямд)
мах - 25000₮ (ҮНЭТЭЙ ⚠️)
давс - 1500₮ (хямд)
```

> 💡 Хоёр list-ийг зэрэг гүйлгэхийн тулд **индексээр** давтана:
> ```python

>     name = products[i]

````python
# Энд бичнэ үү
def print_price_check(products, prices, limit):
    for product, price in zip(products, prices):
        if price > limit:
            print(f"{product}: {price} - Expensive")
        else:
            print(f"{product}: {price} - Cheap")
products = ["сүү", "талх", "мах", "давс"]
prices = [3500, 2000, 25000, 1500]
print_price_check(products, prices, 5000)
````

### B10 — Температурын тайлан (ангилал + тоолуур)

`print_temperature_report(temps)` функц бичээрэй. Температур бүрийг ангилаад, **төгсгөлд нь дүгнэлт** хэвлэнэ.

| Температур | Ангилал |
|---|---|
| 20-аас дээш | Дулаан |
| 0 – 20 | Сэрүүн |
| 0-аас доош | Хүйтэн |

```python
print_temperature_report([25, -5, 12, 30, -12, 3])
```

**Хүлээгдэх үр дүн:**
```
25°C → Дулаан
-5°C → Хүйтэн
12°C → Сэрүүн
30°C → Дулаан
-12°C → Хүйтэн
3°C → Сэрүүн

--- Дүгнэлт ---
Дулаан өдөр: 2
Сэрүүн өдөр: 2
Хүйтэн өдөр: 2
```

> 💡 Гурван тоолуурыг loop-ийн **гадна** 0-ээр эхлүүлээрэй.

````python
# Энд бичнэ үү
def print_temperature_report(temps):
    for temp in temps:
        if temp > 20:
            print(f"{temp}°C: Hot")
        elif temp > 0:
            print(f"{temp}°C: Warm")
        else:
            print(f"{temp}°C: Cold")
print_temperature_report([25, -5, 12, 20, -12, 3])
````

### C — Сурагчийн тайлан

`print_student_report(names, scores, pass_score)` функц бичээрэй. Дараах бүгдийг хийнэ:

1. Сурагч бүрийн нэр, оноо, тэнцсэн эсэхийг хэвлэх
2. Тэнцсэн болон унасан сурагчийн тоог тоолох
3. Ангийн дундаж оноог тооцох
4. Хамгийн өндөр оноотой сурагчийн **нэрийг** олох
5. Төгсгөлд дүгнэлт хэвлэх

```python
names = ["Болд", "Сараа", "Тэмүүлэн", "Ану"]
scores = [85, 45, 92, 58]
print_student_report(names, scores, 60)
```

> 💡 Хамгийн өндөр оноог олохдоо **нэр ба оноог хамт** хадгалж явна:
> ```python
> best_name = names[0]
> best_score = scores[0]
> ```

````python
# Энд бичнэ үү
def print_student_report(names,scores,pass_scores):
    passed = 0
    failed = 0
    for name, score in zip(names, scores):
        if score >= pass_scores:
            print(f"{name}: {score} - Pass")
            passed += 1
        else:
            print(f"{name}: {score} - Fail")
            failed += 1
    print(f"Total Passed: {passed}")
    print(f"Total Failed: {failed}")
    print(f"Total Students: {len(names)}")
    print(f"Class average: {sum(scores)/len(scores)}")
    print(f"Highest score: {names[scores.index(max(scores))]} ({max(scores)})")
names = ["Болд", "Сараа", "Тэмүүлэн", "Ану"]
scores = [85, 45, 92, 58]
print_student_report(names, scores, 60)

````

# 📝 Гэрийн даалгавар

**1. `print_shopping_list(items, quantities)`**
Барааны нэр ба тоо ширхгийн 2 list авна. Бараа бүрийг хэвлээд, тоо ширхэг нь **5-аас их** бол ард нь `(бөөнөөр)` гэж нэмнэ. Төгсгөлд нийт барааны тоог хэвлэнэ.

**2. `print_word_analysis(words)`**
Үгсийн list авна. Үг бүрийн ард түүний уртыг хэвлэнэ. Дараа нь дараах дүгнэлтийг гаргана:
- Хамгийн урт үг
- Хамгийн богино үг
- Үгсийн дундаж урт

**3. `print_countdown(start)`**
Тооноос эхлээд 1 хүртэл буцааж тоолно. Гэхдээ:
- Тэгш тоо бол `"{n} (тэгш)"`
- Сондгой тоо бол `"{n} (сондгой)"`
- Төгсгөлд `"Дууслаа! 🚀"`

> 💡 Буцаж тоолохдоо `range(start, 0, -1)` ашиглана.

**4. 🚀 Нэмэлт — `print_budget(categories, amounts, budget)`**
Зардлын ангилал ба дүнгийн 2 list, нийт төсөв авна.
- Ангилал бүрийг хэвлэж, нийт төсвийн хэдэн хувийг эзэлж байгааг харуулна
- Нийт зардлыг тооцоолно
- Төсвөөс хэтэрсэн бол `"⚠️ Төсөв хэтэрсэн!"`, үгүй бол үлдэгдлийг хэвлэнэ

````python
# Даалгавар 1
def print_shopping_list(items,quantities):
    for item, quantity in zip(items, quantities):
        if(quantity > 5):
            print(f"{item}: {quantity} - Boonoor")
        else:
            print(f"{item}: {quantity}")
items = ["сүү", "талх", "мах", "давс"]
quantities = [2, 5, 3, 7]
print_shopping_list(items, quantities)
````

````python
# Даалгавар 2
def print_word_analysis(words):
    longest_word = words[0]
    shortest_word = words[0]
    sum_length = 0
    for word in words:
        length = len(word)
        print(f"{word}: Length = {length} ")
        if length > len(longest_word):
            longest_word = word
        if length < len(shortest_word):
            shortest_word = word
        sum_length += length
    average_length = sum_length / len(words)
    print(f"Longest word: {longest_word}")
    print(f"Shortest word: {shortest_word}")
    print(f"Average length: {average_length}")


words = ["python", "data", "machine", "learning"]
print_word_analysis(words)
````

````python
# Даалгавар 3
def print_countdown(start):
    for i in range(start, 0, -1):
        if(i % 2 == 0):
            print(f"{i} even")
        else:
            print(f"{i} odd")   
    print("Finished!")
print_countdown(10)
````

````python
# Даалгавар 4 (нэмэлт)
def print_budget(categories,amounts,budget):
    total = 0
    for category, amount in zip(categories, amounts):
        total += amount
        print(f"{category}: {amount/budget*100:.2f}% of budget")
    print(f"Total: {total}")
    print(f"Percentage of budget: {total/budget*100:.2f}%")
    if total > budget:
        print("Over budget")
    elif total < budget:
        print("Under budget")
    else:
        print("On budget")
categories = ["Food", "Transport", "Entertainment"]
amounts = [200, 150, 100]
budget = 500
print_budget(categories, amounts, budget)
````

```python
def show(items):
    total = 0                    # ← loop-ийн ГАДНА эхлүүлнэ
    for item in items:
        total = total + item     # ← loop-ийн ДОТОР нэмнэ
    print(total)                 # ← loop дууссаны ДАРАА хэвлэнэ
```

### 4️⃣ Аварга хайх (Find best)
```python
def show(items):
    best = items[0]              # ← эхний утгыг "аварга" болгоно
    for item in items:
        if item > best:
            best = item          # ← илүү сайн нь тааралдвал солино
    print(best)
```
