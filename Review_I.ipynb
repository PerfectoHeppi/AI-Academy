# 1. "Алдаа гарна" гэдэг ойлголт

`10 == "10"` эсвэл `"c" in d` гэсэн код алдаа гаргах уу?

Үгүй. Эдгээр нь **True эсвэл False** буцаана.

## Дүрэм

Харьцуулах оператор (`==`, `!=`, `>`, `<`, `in`) нь **үргэлж True эсвэл False** буцаана.
Ямар ч хоёр утгыг харьцуулж болно, алдаа гарахгүй.

````python
# Эхлээд таамагла: юу хэвлэгдэх вэ?
# Дараа нь ажиллуул

print(10 == "10")
````

**Тайлбар:** `10` бол тоо, `"10"` бол текст. Өөр төрөл тул тэнцүү биш.
Гэхдээ алдаа гарахгүй, зүгээр л `False` буцаана.

````python
d = {"a": 1, "b": 2}

print("c" in d)   # d дотор "c" гэсэн түлхүүр байна уу?
print("a" in d)   # d дотор "a" гэсэн түлхүүр байна уу?
````

**Тайлбар:** `in` нь "байна уу" гэж асууна. Хариу нь True эсвэл False.

````python
# not оператор утгыг эсрэгээр болгоно

print(5 > 3)          # энэ юу вэ?
print(not (5 > 3))    # эсрэгээр нь юу вэ?
````

**Тайлбар:** `5 > 3` бол True. `not True` нь **False**.

# 2. `append` ба `extend`

Эдгээр хоёр нь адилхан харагддаг ч огт өөр ажиллана.

## Асуулт

```python
nums = [1, 2, 3]
nums.append([4, 5])
len(nums)  # хэд вэ?
```

Ажиллуулж үзье.

````python
nums = [1, 2, 3]
nums.append([4, 5])

print(nums)
print(len(nums))
````

**Тайлбар:** `append` нь өгсөн зүйлийг **бүхэлд нь нэг элемент** болгож нэмнэ.

`[4, 5]` гэсэн жагсаалт бүхэлдээ 4 дэх элемент болсон. Тэгэхээр урт нь **4**.

## Харин элементүүдийг тус тусад нь нэмэх бол

````python
nums = [1, 2, 3]
nums.extend([4, 5])

print(nums)
print(len(nums))
````

````python
# += нь extend-тэй адилхан

nums = [1, 2, 3]
nums += [4, 5]

print(nums)
print(len(nums))
````

## Хураангуй

| Команд | Үр дүн | Урт |
|---|---|---|
| `nums.append([4,5])` | `[1, 2, 3, [4, 5]]` | 4 |
| `nums.extend([4,5])` | `[1, 2, 3, 4, 5]` | 5 |
| `nums += [4,5]` | `[1, 2, 3, 4, 5]` | 5 |

**Санах арга:** append = "нэг зүйл нэм", extend = "сунга"

### Өөрөө туршиж үз

````python
# Доорх кодыг ажиллуулж, ялгааг хар

a = ["Бат", "Оюу"]
b = ["Бат", "Оюу"]

a.append(["Дорж", "Сараа"])
b.extend(["Дорж", "Сараа"])

print("append:", a, "урт:", len(a))
print("extend:", b, "урт:", len(b))
````

# 3. `/` ба `//`, float гэж юу вэ

`12 / 3` нь 4 юу, эсвэл 4.0 юу? Хариу нь **4.0**.

## Дүрэм

- `/` нь **үргэлж float** (бутархай) буцаана, хуваалт тэгш байсан ч
- `//` нь **бүхэл хуваалт**, бутархай хэсгийг хаяна

````python
print(12 / 3)     # юу гарах вэ?
print(12 // 3)    # юу гарах вэ?
````

````python
print(7 / 2)      # энгийн хуваалт
print(7 // 2)     # бүхэл хуваалт
````

````python
# type() нь төрлийг харуулна

print(type(12 / 3))
print(type(12 // 3))
print(type(3.0))
print(type(3))
````

**Тайлбар:** `3.0` дээр цэг байгаа тул **float**. `3` бол **int**.

Шалгалтад "3.0 ямар төрөлд хамаарах вэ" гэсэн асуулт байсан. Хариу нь **float**.

````python
# % нь үлдэгдэл гаргана. Тэгш сондгой ялгахад хэрэглэнэ

print(7 % 2)      # 7-г 2-т хуваахад үлдэгдэл
print(8 % 2)      # 8-г 2-т хуваахад үлдэгдэл

print(7 % 2 == 1)  # сондгой юу?
print(8 % 2 == 0)  # тэгш юу?
````

# 5. `while` давталт хэзээ зогсох вэ

## Асуулт

```python
x = 10
while x > 0:
    x = x - 3
```

Давталт дуусахад x хэд вэ?

````python
x = 10
while x > 0:
    print("x одоо:", x, "-> 3 хасна")
    x = x - 3

print("Давталт дууслаа. x =", x)
````

**Тайлбар:** x = 1 байх үед `1 > 0` нөхцөл **дахин биелнэ**, тул нэг удаа илүү хасна.

10 → 7 → 4 → 1 → **-2**

x нь -2 болмогц `-2 > 0` худал болж давталт зогсоно.

**Санах зүйл:** нөхцөл биелэх бүрд давталт дахин ажиллана. Сүүлийн эргэлтийг мартаж болохгүй.

# 6. Trace: хувьсагчийн утгыг мөшгих

## Асуулт

```python
a = 2
b = 5
a = a + b
b = a - b
print(a, b)
```

## Гол зарчим

Мөр бүр **дараалан** гүйцэтгэгдэнэ. Дараагийн мөр өмнөх мөрийн **шинэ утгыг** ашиглана.

````python
a = 2
print("Мөр 1: a =", a)

b = 5
print("Мөр 2: b =", b)

a = a + b
print("Мөр 3: a = a + b =", a, "  (2 + 5)")

b = a - b
print("Мөр 4: b = a - b =", b, "  ЭНД a нь аль хэдийн 7 болсон! 7 - 5 = 2")

print()
print("Эцсийн хариу:", a, b)
````

## Хүснэгтээр

| Ажиллах мөр | a | b |
|---|---|---|
| `a = 2` | 2 | |
| `b = 5` | 2 | 5 |
| `a = a + b` | **7** | 5 |
| `b = a - b` | 7 | **2** |
| `print(a, b)` | 7 | 2 |

**Хамгийн олон гардаг алдаа:** `b = a - b` дээр a-г 2 гэж бодох. Үгүй, тэр үед a нь аль хэдийн **7** болсон.

## Дасгал: өөрөө хий

Доорх кодыг **ажиллуулахаас өмнө** хүснэгт дээр бөглөж таамагла.

````python
# Таамаглал бич: p = ?  q = ?

p = 4
q = 3
p = p * q
q = p - q
print(p, q)
````

````python
# Таамаглал бич: total = ?

total = 0
for n in [3, 5, 2]:
    if n > 2:
        total = total + n
        print("n =", n, "-> total одоо:", total)
    else:
        print("n =", n, "-> нөхцөл биелэхгүй, total хэвээр:", total)

print("Эцсийн total:", total)
````
