## Exercise 1 — Palindrome Checker 🔁

Хэрэглэгчээс үг эсвэл өгүүлбэр асууж, **урвуу уншихад ижил** эсэхийг шалга.

- Зай, том/жижиг үсгийг тооцохгүй
- `racecar`, `level`, `Never odd or even` → палиндром

**Жишээ гаралт:**
```
Enter a word: Racecar
Yes, it is a palindrome! 🔁
```

💡 `.lower()` , `.replace(" ", "")` , `[::-1]`

````python
# Exercise 1
# Кодоо энд бичнэ үү
def Palindrome(Word):
    Word = Word.lower()
    Word = Word.replace(" ", "")
    if(Word == Word[::-1]):
        print("Word is palindrome")
    else:
        print("word is not palindrome")
Palindrome("Never odd or eve")
````

## Exercise 2 — Initials Generator

Хэрэглэгчээс бүтэн нэр асууж, **эхний үсгүүдээр** товчлол хий.

**Жишээ гаралт:**
```
Enter your full name: bat erdene dorj
Initials: B.E.D.
```

💡 `.split()` , `for` , `word[0].upper()`

````python
# Exercise 2
# Кодоо энд бичнэ үү
def Initial_Generator():
    name = input("Enter your full name: ")
    name = name.split()
    initials = ""
    for n in name:
        initials += n[0].upper() + "."
    print(initials)

Initial_Generator()
````

## Exercise 3 — Vowel Remover

Хэрэглэгчийн өгүүлбэрээс **бүх эгшгийг ( a e i o u ) хасаад** хэвлэ.

**Жишээ гаралт:**
```
Enter a sentence: I love Python
Without vowels:  lv Pythn
```

💡 Шинэ текстийг `result = ""` -ээс эхлээд `for` дотор нэмж цуглуулна.

````python
# Exercise 3
# Кодоо энд бичнэ үү
def vowel_remover(sentence):
    vowels = "aeiouy"
    result = ""
    for n in sentence:
        for x in n:
            if x not in vowels:
                result += x
                
    print(result)
vowel_remover("Tugsbayar Erdenebaatar")
````

## Exercise 4 — Longest Word Finder

Өгүүлбэр доторх **хамгийн урт үгийг** ол, түүний уртыг нь хамт хэвлэ.

**Жишээ гаралт:**
```
Enter a sentence: Python makes programming easy
Longest word: programming (11 letters)
```

💡 `longest = ""` гэж эхлээд `if len(word) > len(longest):` гэж харьцуулна.

````python
# Exercise 4
# Кодоо энд бичнэ үү
def longest_word_finder(sentence):
    longest = ""
    sentence = sentence.split()
    for x in sentence:
        if(len(x) > len(longest)):
            longest = x
    print(longest)
longest_word_finder("I love you you little snowflake")
````

## Exercise 5 — Leetspeak Translator 1337

Текстийг "хакер бичиг" болгож хөрвүүл:

| Үсэг | Болох нь |
|---|---|
| a | 4 |
| e | 3 |
| i | 1 |
| o | 0 |
| s | 5 |

Бусад тэмдэгт хэвээрээ.

**Жишээ гаралт:**
```
Enter text: I love Python and AI
Leet: 1 l0v3 pyth0n 4nd 41
```

💡 `for ch in text.lower():` дотор `if` / `elif` гинжин шалгалт.

````python
def Leetspeak_Translater(sentence):
    for x in sentence:
        if(x == "a"):
            sentence = sentence.replace("a","4")
        elif(x == "e"):
            sentence = sentence.replace("e","3")
        elif(x == "i"):
            sentence = sentence.replace("i","1")
        elif(x == "o"):
            sentence = sentence.replace("o","0")
        elif(x == "s"):
            sentence = sentence.replace("s","5")
    print(sentence)
Leetspeak_Translater("I love Python and AI savage")
````
